```python
import os, math, time, random
from pathlib import Path
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import torch.optim as optim
from torchvision import datasets, transforms, utils as vutils
from PIL import Image, ImageFilter
from tqdm import tqdm
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
class PrepareDataset(Dataset):
    """Custom dataset for loading aligned image pairs (A|B) from the
    edges2shoes dataset.
    Each image is split into left (A: edges) and right (B: photo) halves.
    """
    def __init__(self, root, split="train", size=256, hflip=True,
        in_ch="auto"):
        self.root = Path(root) / split
        self.paths = sorted([p for p in self.root.iterdir() if
        p.suffix.lower() in {".jpg", ".png", ".jpeg"}])
        self.size = size
        self.hflip = hflip # whether to apply random horizontal flip to both A and B
        self.in_ch = in_ch # 1 or 3
        # Transforms: resize to common size, convert to tensor, normalizto [-1,1]
        self.resize = transforms.Resize((size, size),
        interpolation=Image.BICUBIC)
        self.to_tensor = transforms.ToTensor()
        self.norm1 = transforms.Normalize([0.5], [0.5]) # for 1-channel
        input
        self.norm3 = transforms.Normalize([0.5, 0.5, 0.5], [0.5, 0.5,
        0.5]) # for 3-channel input

    def __len__(self):
        return len(self.paths)
    
    def _split_ab(self, img):
        w, h = img.size
        # Split image into left and right halves
        A = img.crop((0, 0, w//2, h)) # left half: edges
        B = img.crop((w//2, 0, w, h)) # right half: photo
        return A, B

    def __getitem__(self, idx):
        # Load image and ensure it's in RGB mode
        img = Image.open(self.paths[idx]).convert("RGB")
        A_img, B_img = self._split_ab(img)
        # Resize both halves to the target size
        A_img = self.resize(A_img)
        B_img = self.resize(B_img)
        # Random horizontal flip (applied to both A and B together)
        if self.hflip and random.random() < 0.5:
            A_img = A_img.transpose(Image.FLIP_LEFT_RIGHT)
            B_img = B_img.transpose(Image.FLIP_LEFT_RIGHT)
        # Determine number of input channels for A
        if self.in_ch == 1:
        # Convert A to grayscale if 1-channel input is desired
            A_t = transforms.functional.rgb_to_grayscale(self.to_tensor(A_img),
            num_output_channels=1)
            A_t = self.norm1(A_t)
        else:
        # Use 3 channels for A (even if it's originally an edge image)
            A_t = self.norm3(self.to_tensor(A_img))
        # Convert B to tensor and normalize (always 3-channel RGB output)
        B_t = self.norm3(self.to_tensor(B_img))
        return A_t, B_t
```

```python
# Initialize datasets and loaders
root = "./edges2shoes"
train_ds = PrepareDataset(root, split="train", size=256, hflip=True,
in_ch=3) # use 3-channel edges
val_ds = PrepareDataset(root, split="val", size=256, hflip=False,
in_ch=3)
train_loader = DataLoader(train_ds, batch_size=8, shuffle=True,
num_workers=2, pin_memory=True)
val_loader = DataLoader(val_ds, batch_size=8, shuffle=False,
num_workers=2, pin_memory=True)
print("Dataset sizes:", len(train_ds), "training images,", len(val_ds),
"validation images.")
```

```python
# Building blocks for the U-Net generator:
def conv_block(in_c, out_c, k=4, s=2, p=1, norm=True, activation='lrelu'):
    """Args:
    in_c, out_c: input/output channels
    k,s,p: kernel, stride, padding
    norm: whether to use InstanceNorm
    activation function: 'lrelu' or 'relu'
    A convolutional block: Conv2d -> (InstanceNorm) -> LeakyReLU/ReLU
    """
    layers = [nn.Conv2d(in_c, out_c, kernel_size=k, stride=s, padding=p,
    bias=not norm)]
    if norm:
        layers.append(nn.InstanceNorm2d(out_c))
    if activation == 'lrelu':
        layers.append(nn.LeakyReLU(0.2, inplace=True))
    elif activation == 'relu':
        layers.append(nn.ReLU(inplace=True))
    return nn.Sequential(*layers)

def deconv_block(in_c, out_c, k=4, s=2, p=1, dropout=False):
    """A transposed-convolution block: ConvTranspose2d -> InstanceNorm ->
    ReLU -> (Dropout optional)"""
    layers = [
    nn.ConvTranspose2d(in_c, out_c, kernel_size=k, stride=s, padding=p,
    bias=False),
    nn.InstanceNorm2d(out_c),
    nn.ReLU(inplace=True),
    ]
    if dropout:
        layers.append(nn.Dropout(0.5))
    return nn.Sequential(*layers)
```

```python
# U-Net Generator (encoder-decoder with skip connections):
class UNetGenerator(nn.Module):

    def __init__(self, in_ch=3, out_ch=3, nf=64):
        """Args:
        in_ch: number of input channels (1 or 3 for edges)
        out_ch: number of output channels (3 for RGB photo)
        nf: number of filters in the conv layer
        """
        super().__init__()
        # Encoder (Downsampling path)
        self.e1 = conv_block(in_ch, nf, norm=False) # Output: nf features, (H/2 x W/2)
        self.e2 = conv_block(nf, nf*2) # Output: nf*2 features (H/4 x W/4)
        self.e3 = conv_block(nf*2, nf*4) # (H/8 x W/8)
        self.e4 = conv_block(nf*4, nf*8) # (H/16 x W/16)
        self.e5 = conv_block(nf*8, nf*8) # (H/32 x W/32)
        self.e6 = conv_block(nf*8, nf*8, norm=False) # (H/64 x W/64) bottleneck
        # Decoder (Upsampling path with skip connections; dropout on deeper layers)
        self.d1 = deconv_block(nf*8, nf*8, dropout=True) # Upsamplebottleneck (H/32)
        self.d2 = deconv_block(nf*16, nf*8, dropout=True) # concat with e5 -> (H/16)
        self.d3 = deconv_block(nf*16, nf*4) # concat with e4 -> (H/8)
        self.d4 = deconv_block(nf*8, nf*2) # concat with  e3 -> (H/4)
        self.d5 = deconv_block(nf*4, nf) # concat withe2 -> (H/2)
        self.d6 = nn.Sequential( # concat with e1 -> (H x W)
        nn.ConvTranspose2d(nf*2, out_ch, kernel_size=4, stride=2,
        padding=1),
        nn.Tanh())

    def forward(self, x):
        # Encoder forward pass
        e1 = self.e1(x) # e1: first conv output
        e2 = self.e2(e1)
        e3 = self.e3(e2)
        e4 = self.e4(e3)
        e5 = self.e5(e4)
        e6 = self.e6(e5) # e6: bottleneck features
        # Decoder forward pass with skip connections
        d1 = self.d1(e6) # upsample bottleneck
        d1 = torch.cat([d1, e5], dim=1) # concat skip from e5
        d2 = self.d2(d1)
        d2 = torch.cat([d2, e4], dim=1) # skip from e4
        d3 = self.d3(d2)
        d3 = torch.cat([d3, e3], dim=1) # skip from e3
        d4 = self.d4(d3)
        d4 = torch.cat([d4, e2], dim=1) # skip from e2
        d5 = self.d5(d4)
        d5 = torch.cat([d5, e1], dim=1) # skip from e1
        out = self.d6(d5) # final output layer
        return out
```

```python
# PatchGAN Discriminator:
class PatchDiscriminator(nn.Module):
    def __init__(self, in_ch=6, nf=64):
        super().__init__()
        # No normalization on first layer per GAN convention
        self.c1 = conv_block(in_ch, nf, norm=False) # Input: 6-ch -> 64 filters, (H/2 x W/2)
        self.c2 = conv_block(nf, nf*2) # 64 -> 128 filters, (H/4 x W/4)
        self.c3 = conv_block(nf*2, nf*4) # 128 -> 256 filters, (H/8 x W/8)
        self.c4 = conv_block(nf*4, nf*8, s=1) # 256 -> 512 filters, (H/8 x W/8) <-- stride 1 (no further spatial downsampling, receptive field ~70x70)
        self.c5 = nn.Conv2d(nf*8, 1, kernel_size=4, stride=1, padding=1) #  Output: 1 filter, (H/8 x W/8) logits map
        
    def forward(self, a, b):
        # a: input edge image, b: target or generated image
        x = torch.cat([a, b], dim=1) # concatenate along channels -> shape [batch, 6, H, W]
        x = self.c1(x)
        x = self.c2(x)
        x = self.c3(x)
        x = self.c4(x)
        x = self.c5(x) # final output shape: [batch, 1, H/8, W/8]
        return x
```

```python
# Initialize the Generator and Discriminator
G = UNetGenerator().to(device)
D = PatchDiscriminator().to(device)
# Loss functions: adversarial loss (GAN) and L1 loss
criterion_gan = nn.BCEWithLogitsLoss()
criterion_l1 = nn.L1Loss()
# Optimizers for G and D
lr = 2e-4
beta1, beta2 = 0.5, 0.999
opt_G = optim.Adam(G.parameters(), lr=lr, betas=(beta1, beta2))
opt_D = optim.Adam(D.parameters(), lr=lr, betas=(beta1, beta2))
# Weight for L1 loss term
lambda_L1 = 100.0 # L1 weight
```

```python
# Helper function to save a grid of samples:
def save_sample(a, fake_b, out_path, b=None):
    """
    Save a grid: [A (edge), B (real), G(A) (fake)] for the first 8
    samples.
    Tensors expected in [-1,1].
    """
    a_ = (a[:8].cpu() + 1) / 2.0
    f_ = (fake_b[:8].detach().cpu() + 1) / 2.0
    if b is not None:
        b_ = (b[:8].cpu() + 1) / 2.0
        grid = vutils.make_grid(torch.cat([a_, b_, f_], dim=0), nrow=8,
        padding=2)
    else:
        grid = vutils.make_grid(torch.cat([a_, f_], dim=0), nrow=8,
        padding=2)
        vutils.save_image(grid, out_path)
```

```python
# Training loop
epochs = 12 # number of epochs to train (can adjust as needed)
outdir = Path("./pix2pix_runs")
outdir.mkdir(parents=True, exist_ok=True)

for epoch in range(1, epochs+1):
    G.train()
    D.train()
    pbar = tqdm(train_loader, desc=f"Epoch {epoch}/{epochs}")
    running_G, running_D = 0.0, 0.0 # accumulate losses for reporting

    for a, b in pbar:
        a, b = a.to(device, non_blocking=True), b.to(device,
        non_blocking=True)
        # Update Discriminator (maximize ability to distinguish real vs fake):
        opt_D.zero_grad(set_to_none=True)
        # Train on real batch
        pred_real = D(a, b) # D's predictions for real pair (a,b)
        valid = torch.ones_like(pred_real, device=device) # label = 1 for all patches (real)
        loss_D_real = criterion_gan(pred_real, valid) # how far were real pairs from being classified as real (want this low)
        
        # Train on fake batch
        with torch.no_grad():
            fake_b = G(a) # generate fake image from a (detached G for D update)
        pred_fake = D(a, fake_b) # D's predictions for fake pair (a, G(a))
        fake = torch.zeros_like(pred_fake, device=device) # label = 0 for fake patches
        loss_D_fake = criterion_gan(pred_fake, fake) # how far were fake pairs from being classified as fake (want this low)
        # Combine discriminator loss and update
        loss_D = 0.5 * (loss_D_real + loss_D_fake) # total D loss (scaled by 0.5 to slow down D learning relative to G)
        loss_D.backward() # compute gradients for D
        opt_D.step() # update D's weights
        # Update Generator (maximize fooling the discriminator + L1 similarity):
        opt_G.zero_grad(set_to_none=True)
        fake_b = G(a) # generate fake image (this time, we'll keep grad for G)
        pred_fake_for_G = D(a, fake_b) # D's prediction on the fake (a, G(a))
        loss_G_adv = criterion_gan(pred_fake_for_G, valid) # generator tries to make D output "valid (1)" for fakes
        loss_G_L1 = criterion_l1(fake_b, b) * lambda_L1 # L1 pixel loss * 100
        loss_G = loss_G_adv + loss_G_L1
        loss_G.backward() # compute gradients for G
        opt_G.step() # update G's weights
        # Accumulate losses for monitoring
        running_G += loss_G.item()
        running_D += loss_D.item()
        pbar.set_postfix(G=running_G/(pbar.n+1), D=running_D/(pbar.n+1))

    # End of epoch: evaluate on val set and save sample output:
    G.eval()
    with torch.no_grad():
        # Get a batch from validation set
        a_val, b_val = next(iter(val_loader))
        a_val, b_val = a_val.to(device), b_val.to(device)
        fake_val = G(a_val) # generator output on val inputs
        # Save a sample grid of inputs and outputs (you could also include b_val to compare)
        # save_sample(a_val, b_val, fake_val, outdir / f"epoch_{epoch:03d}.png")
        save_sample(a_val, fake_val, outdir / f"epoch_{epoch:03d}.png")
        # (Optional) Save model checkpoints after each epoch
        torch.save(G.state_dict(), outdir / f"G_epoch_{epoch:03d}.pt")
        torch.save(D.state_dict(), outdir / f"D_epoch_{epoch:03d}.pt")
    print("Training finished. Check samples in", str(outdir))
```
